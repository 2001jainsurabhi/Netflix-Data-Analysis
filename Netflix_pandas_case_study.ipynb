### 16. What is the average duration of movies in minutes?


```python
# Code here
movie_df = df[df['type'] == 'Movie']
movie_df.shape
```

```python
movie_df.reset_index()
```

```python
movie_df.reset_index(drop = True, inplace = True)
```

```python
movie_df['duration'][0]
```

```python
int('90 min'.split()[0])
```

```python
def extract_numeric(s):
  return int(s.split()[0])

extract_numeric('90 min')
```

```python
movie_df1 = movie_df.dropna(subset = ['duration'])
```

```python
movie_df1['duration_numeric'] = movie_df1['duration'].apply(extract_numeric)
# movie_df['duration_numeric'] = movie_df['duration'].apply(lambda x: int(x.split()[0]))
movie_df1['duration_numeric']
```

```python
movie_df1['duration_numeric'].mean()
```

```python
# release_year wise average duration
movie_df1.groupby('release_year').agg({'duration_numeric':['min', 'max', 'mean'], 'show_id':'nunique'}).reset_index()
```
